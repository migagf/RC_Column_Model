```python
# Model Tester for YHY

# Simple python script to thest a Bouc-Wen model
# On my framework I have implementations of 3 bouc-wen models

# 1. Standard Bouc-Wen
# 2. Bouc-Wen with strength and stiffness degradation by Pelliciari (not good)
# 3. Bouc-Wen with strength and stiffness degradation by Kottari (better)

from material_models import *
import matplotlib.pyplot as plt

# material_models comes with a tester function
```

```python
# ::: Standard Bouc-Wen :::
# Regular Bouc-Wen model
parameters = [
    0.1, # alpha
    0.2, # gamma
    1.0, # beta
    5.0, # n
    10.0  # k
    ]

plastic_hinge = BoucWen(parameters)

# Create a strain vector
strains = createCyclicStrain(maxStrains=[0.1, 1.0, 2.0, 3.0], dStrain=0.01)

# Test the material
strain_vec, stress_vec = material_test(plastic_hinge, strains[0])

# Plot the results
plt.plot(strain_vec, stress_vec)
plt.title("Standard Bouc-Wen Model")
plt.xlabel("Strain")
plt.ylabel("Stress")
plt.grid()
plt.show()
```

```python
# ::: Kottari Bouc-Wen :::

parameters = [
    0.5,  # eta1
    4.0,  # k0
    4.0,  # sy0
    0.25,  # sig
    0.25,  # lam
    2.0,  # mup
    1.0,  # sigp
    0.6,  # rsmax
    2.0,  # n
    -0.05,  # alpha
    2.0,  # alpha1
    0.5,  # alpha2
    0.001   # betam1
]

plastic_hinge = DegradingBoucWen(parameters)

# Create a strain vector
strains = createCyclicStrain(maxStrains=[0.1, 1.0, 2.0, 3.0], dStrain=0.01)
# Test the material
strain_vec, stress_vec = material_test(plastic_hinge, strains[0])

# Plot the results
plt.plot(strain_vec, stress_vec)
plt.title("Kottari Bouc-Wen Model")
plt.xlabel("Strain")
plt.ylabel("Stress")
plt.grid()
plt.show()
```
