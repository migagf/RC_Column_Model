```python
import numpy as np
import scipy as sp
```

```python
# Parameters of the reinforced concrete column

# Column external diameter
D = 50
# Clear cover
cc = 5

# Compression zone size
c = 25

alpha = np.arccos((D/2 - c)/(D/2 - cc))

# I'm doing this to find the yield rotation
r = D/2 - cc
yp = 2 * r * np.cos(alpha) / (np.pi + 2 * alpha)

d = yp + D/2

print('Depth to centroid of reinf:', d)
```

```python
eps_s = 0.002
eps_c = (d - c) / c * eps_s

print('Concrete strain', eps_c)
```
